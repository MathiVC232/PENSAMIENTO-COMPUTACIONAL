```
#1.  Sumar 5 números y mostrar el resultado.
suma = 0
for i in range (1,6):
  num = float(input(f"Ingrese el número {i}: "))
  suma = suma + num

print(f"El resultado de la suma de los 5 números es: {suma}")
```

```
#2. Permitir ingresar N números y sumarlos. Mostrar resultado.
suma = 0
n = int(input("Ingrese la cantidad de valores que quiere sumar: "))

for i in range (n):
  num = float(input(f"Ingrese el número {i}: "))
  suma = suma + num
print(f"La suma de los números es: {suma}")
```

```
#3. Mostrar el residuo de una división.

dvdno = float(input("Ingrese un dividendo: "))
dvsor = float(input("Ingrese un divisor: "))

residuo = dvdno % dvsor

print(f"El residuo de la división de {dvdno} entre {dvsor} es de: {residuo}")
```

```
#4. Sumar los números enteros múltiplos de 3, comprendidos entre el 1 y el 50. Mostrar en pantalla los números y finalmente la sumatoria.
suma = 0
for i in range (1,50):
  if i % 3 == 0:
    print(i)
    suma = suma + i

print(f"El resultado de la suma es: {suma}")
```

```
#5. Calcular el área de un cuadrado.

lado  = float(input("Ingrese el valor del lado de su cuadrado: "))

area = lado * lado

print(f"El área de su cuadrado es {area} cm2: ")
```

```
#6. Calcular el área de un rectángulo.

base = float(input("Ingrese el valor de la base de su rectángulo: "))
altura = float(input("Ingrese el valor de la altura de su rectángulo: "))

area = base * altura

print(f"El área de su rectángulo es igual a {area}cm2")
```

```
#7. Calcular el IVA de un producto.

iva = float(input("Ingrese el valor IVA de su país: "))
producto = float(input("Ingrese el valor de su producto: "))

iva_calculado = (producto * (iva/100))
total_pagar = producto + iva_calculado
print("\n --FACTURA DE PRODUCTO--")
print(f"El valor de su producto es: ${producto}")
print(f"El valor del iva de su producto es: ${iva_calculado}")
print(f"El valor total a pagar con iva añadido es: ${total_pagar}")
```

```
#8. Ingresar el valor de un producto incluido el IVA y que el dfd muestre el valor sin impuesto.
precio  = float(input("Ingrese el precio de su producto: "))
porcen = float(input("Ingrese el porcentaje de IVA que desea utilizar: "))

impuesto = porcen / 100
precio_sin_iva = precio / (1 + impuesto)
iva = precio - precio_sin_iva

print("\n --FACTURA DE PRODUCTO--")
print(f"El valor de su producto es: ${precio}")
print(f"El valor del iva de su producto es: ${iva}")
print(f"El valor total a pagar sin IVA añadido es: ${precio_sin_iva}")
```

```
#Verificar si el número ingresado es positivo o negativo

num = float(input("Ingrese el número que desea analizar: "))

if num > 0:
  print(f"El número {num} es positivo")
elif num < 0:
  print(f"El número {num} es negativo")
else:
  print(f"El número {num} es cero y se lo considera neutro")
```

```
#Determinar cuánto debe pagar un cliente por el estacionamiento de su vehículo, considere que el valor es de $1 por hora o fracción. Se conoce la placa y el tiempo de estacionamiento en horas y minutos
placa = input("Ingrese la placa del vehículo: ")

horas = int(input("Ingrese las horas de estacionamiento: "))
minutos = int(input("Ingrese los minutos de estacionamiento: "))

if horas > 0:
    total_pagar = horas + 1
elif minutos > 0:
    total_pagar = horas
else:
    total_pagar = horas
print("\n--- FACTURA DE ESTACIONAMIENTO ---")
print("Placa:", placa)
print("Tiempo:", horas, "horas y", minutos, "minutos")
print("Valor a pagar: $", total_pagar)
```
